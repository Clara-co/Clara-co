```python
from huggingface_hub import HfApi
import pandas as pd

# Initialize API
api = HfApi()

# Health-related search keywords
keywords = [
    "health", "medical", "clinical", "medicine", "disease",
    "treatment", "diagnosis", "healthcare", "hospital", "patient"
]

# Function to collect models with parameters
def collect_models_with_parameters(api, query_str):
    df = pd.DataFrame(columns=['id', 'tags', 'downloads', 'parameters', 'category'])
    models = api.list_models(search=query_str)
    for model in models:
        # Skip if required fields are missing
        if not model.id or not model.tags or model.downloads is None:
            continue
        new_row = {
            'id': model.id,
            'tags': model.tags,
            'downloads': model.downloads,
            'parameters': getattr(model, 'model_parameters', None),  # May be None if not available
            'category': query_str
        }
        df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
    return df

# Collect and combine all models
all_dfs = [collect_models_with_parameters(api, kw) for kw in keywords]
combined_df = pd.concat(all_dfs, ignore_index=True)

# Remove duplicates
combined_df.drop_duplicates(subset='id', inplace=True)

# Save to Excel
with pd.ExcelWriter("health_models_with_parameters.xlsx", engine='xlsxwriter') as writer:
    # Header note
    header_note = pd.DataFrame({
        "Note": ["This spreadsheet lists Hugging Face models related to health domains, including their download counts, associated tags, and model parameters."]
    })
    header_note.to_excel(writer, sheet_name="Health Models", index=False, startrow=0)

    # Write model data
    combined_df.to_excel(writer, sheet_name="Health Models", index=False, startrow=3)

print("Saved spreadsheet as 'health_models_with_parameters.xlsx'")
```

```python
from huggingface_hub import HfApi
import pandas as pd

# Initialize Hugging Face API
api = HfApi()

# Define health-related keywords
keywords = [
    "health", "medical", "clinical", "medicine", "disease",
    "treatment", "diagnosis", "healthcare", "hospital", "patient"
]

# Function to collect models for each keyword
def collect_models(api, keyword):
    df = pd.DataFrame(columns=['id', 'tags', 'downloads', 'category'])
    models = api.list_models(search=keyword)
    for model in models:
        if not model.id or not model.tags or model.downloads is None:
            continue
        row = {
            'id': model.id,
            'tags': model.tags,
            'downloads': model.downloads,
            'category': keyword
        }
        df = pd.concat([df, pd.DataFrame([row])], ignore_index=True)
    return df

# Collect models across all keywords
all_dfs = [collect_models(api, kw) for kw in keywords]
combined_df = pd.concat(all_dfs, ignore_index=True)
combined_df.drop_duplicates(subset='id', inplace=True)

# One-hot encode tags efficiently
all_tags = sorted(set(tag for tags in combined_df['tags'] for tag in tags))
tag_matrix = pd.DataFrame([
    {tag: 1 if tag in tags else 0 for tag in all_tags}
    for tags in combined_df['tags']
])
tag_matrix.index = combined_df.index
combined_df = pd.concat([combined_df, tag_matrix], axis=1).copy()

# Save to Excel with header note
with pd.ExcelWriter("health_models_tag_matrix.xlsx", engine="xlsxwriter") as writer:
    header = pd.DataFrame({
        "Note": ["This spreadsheet lists Hugging Face models related to health domains, with binary tag indicators."]
    })
    header.to_excel(writer, sheet_name="Health Models", index=False, startrow=0)
    combined_df.to_excel(writer, sheet_name="Health Models", index=False, startrow=3)

print("✅ Saved spreadsheet as 'health_models_tag_matrix.xlsx'")
```

```python
import requests
from itertools import chain

# Step 1: Define health-related keywords
keywords = [
    "health", "medical", "clinical", "medicine", "disease",
    "treatment", "diagnosis", "healthcare", "hospital", "patient"
]

# Step 2: Collect models by keyword
def collect_models_from_search(api, query_str):
    df = pd.DataFrame(columns=['id', 'tags', 'downloads'])
    models = api.list_models(search=query_str)
    for model in models:
        if not model.id or not model.tags or model.downloads is None:
            continue
        new_row = {
            'id': model.id,
            'tags': model.tags,
            'downloads': model.downloads,
            'category': query_str
        }
        df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
    return df

# Step 3: Combine all keyword searches
all_dfs = [collect_models_from_search(api, kw) for kw in keywords]
combined_df = pd.concat(all_dfs, ignore_index=True).drop_duplicates(subset='id')

# Step 4: Normalize tags
def normalize_tag(tag):
    return tag.strip().lower().replace("-", "_")

combined_df['normalized_tags'] = combined_df['tags'].apply(lambda tags: [normalize_tag(tag) for tag in tags])

# Step 5: Fetch official Hugging Face tags
# Replace this with a real endpoint or static list if needed
reference_tags = {
    "text_classification", "token_classification", "translation", "summarization",
    "question_answering", "image_classification", "speech_recognition", "fill_mask",
    "text_generation", "ner", "medical", "clinical", "healthcare", "multilingual",
    "zero_shot_classification", "sentiment_analysis", "diagnosis", "disease"
}

# Step 6: Validate tags
def validate_tags(tag_list):
    return [tag in reference_tags for tag in tag_list]

combined_df['valid_tags'] = combined_df['normalized_tags'].apply(validate_tags)
combined_df['invalid_tag_count'] = combined_df['valid_tags'].apply(lambda x: x.count(False))

# Step 7: Save to Excel
with pd.ExcelWriter("health_models_clean.xlsx", engine='xlsxwriter') as writer:
    header_note = pd.DataFrame({
        "Note": ["This spreadsheet lists Hugging Face models related to health domains, including their download counts and associated tags."]
    })
    header_note.to_excel(writer, sheet_name="Health Models", index=False, startrow=0)
    combined_df.to_excel(writer, sheet_name="Health Models", index=False, startrow=3)

print("Saved clean spreadsheet with tag validation as 'health_models_clean.xlsx'")
```

```python
# Step 1: Flatten all tags and compute tag-level download totals
from collections import defaultdict

tag_downloads = defaultdict(int)
for _, row in combined_df.iterrows():
    for tag in row['tags']:
        tag_downloads[tag] += row['downloads']

# Step 2: Add tag downloads to each model row
def get_tag_downloads(tag_list):
    return [tag_downloads.get(tag, 0) for tag in tag_list]

combined_df['tag_downloads'] = combined_df['tags'].apply(get_tag_downloads)

# Step 3: Format tags and tag downloads as strings for Excel
combined_df['tags'] = combined_df['tags'].apply(lambda x: ", ".join(x))
combined_df['tag_downloads'] = combined_df['tag_downloads'].apply(lambda x: ", ".join(map(str, x)))

# Step 4: Select and reorder columns
final_df = combined_df[['id', 'tags', 'tag_downloads', 'downloads', 'category']]
final_df.columns = ['model id', 'tags', 'tag downloads', 'model downloads', 'category']

# Step 5: Save to Excel
with pd.ExcelWriter("health_models_clean.xlsx", engine='xlsxwriter') as writer:
    header_note = pd.DataFrame({
        "Note": ["This spreadsheet lists Hugging Face models related to health domains, including their download counts and associated tags."]
    })
    header_note.to_excel(writer, sheet_name="Health Models", index=False, startrow=0)
    final_df.to_excel(writer, sheet_name="Health Models", index=False, startrow=3)

print("Saved clean spreadsheet with tag-level downloads as 'health_models_clean.xlsx'")
```
